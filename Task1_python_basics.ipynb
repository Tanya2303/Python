### Q2:- Write a program that will convert celsius value to fahrenheit.

````python
# Write your code here
temp_c = float(input("Enter the temperature in Celsius: "))
temp_f = (temp_c * 9/5) + 32
print(f"The temperature in Fahrenheit is: {temp_f}")
````

### Q3:- Take 2 numbers as input from the user.Write a program to swap the numbers without using any special python syntax.

````python
# Write your code here
num1 = int(input("Enter the first number: "))
num2 = int(input("Enter the second number: "))
print("Before swapping:")
print("num1 =", num1)
print("num2 =", num2)

temp = num1
num1 = num2
num2 = temp

print("After swapping:")
print("num1 =", num1)
print("num2 =", num2)
````

### Q4:- Write a program to find the euclidean distance between two coordinates.Take both the coordinates from the user as input.

### Q5:- Write a program to find the simple interest when the value of principle,rate of interest and time period is provided by the user.


````python
# Write your code here
principle = float(input("Enter the principle amount: "))
rate = float(input("Enter the rate of interest: "))
time = float(input("Enter the time period: "))
simple_interest = (principle * rate * time) / 100
print(f"The simple interest is: {simple_interest}")
````

### Q6:- Write a program that will tell the number of dogs and chicken are there when the user will provide the value of total heads and legs.

For example:
Input:
heads -> 4
legs -> 12
<br>
Output:
dogs -> 2
chicken -> 2





````python
# Write your code here
heads = int(input("Enter the number of heads: "))
legs = int(input("Enter the number of legs: "))
dogs = (legs - 2 * heads) / 2
chicken = heads - dogs
print(f"dogs -> {dogs}")
print(f"chicken -> {chicken}")
````

### Q7:- Write a program to find the sum of squares of first n natural numbers where n will be provided by the user.

````python
# Write your code here
n = int(input("Enter the number of natural numbers: "))
sum_of_squares = sum(i**2 for i in range(1, n+1))
print(f"The sum of squares of first {n} natural numbers is: {sum_of_squares}")
````

### Q8:- Given the first 2 terms of an Arithmetic Series.Find the Nth term of the series. Assume all inputs are provided by the user.

````python
# Write your code here
a1 = int(input("Enter the first term of the arithmetic series: "))
a2 = int(input("Enter the second term of the arithmetic series: "))
n = int(input("Enter the position of the term to find: "))
d = a2 - a1
nth_term = a1 + (n-1) * d
print(f"The {n}th term of the series is: {nth_term}")
````

### Q9:- Given 2 fractions, find the sum of those 2 fractions.Take the numerator and denominator values of the fractions from the user.

### Q10:- Given the height, width and breadth of a milk tank, you have to find out how many glasses of milk can be obtained? Assume all the inputs are provided by the user.



Input:<br>
Dimensions of the milk tank<br>
H = 20cm, L = 20cm, B = 20cm
<br><br>
Dimensions of the glass<br>
h = 3cm, r = 1cm

````python
# Write your code here
height = int(input("Enter the height of the milk tank: "))
length = int(input("Enter the length of the milk tank: "))
breadth = int(input("Enter the breadth of the milk tank: "))

height_glass = int(input("Enter the height of the glass: "))
radius = int(input("Enter the radius of the glass: "))

volume_milk_tank = height * length * breadth
volume_glass = 3.14 * radius * radius * height_glass

number_of_glasses = volume_milk_tank / volume_glass
print(f"The number of glasses of milk that can be obtained is: {number_of_glasses}")



````
